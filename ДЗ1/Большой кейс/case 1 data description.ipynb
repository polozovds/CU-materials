**Описание таблиц и колонок в кейсе "Падение метрик в кофейне с приложением"**

---

### 1. Таблица: `orders` — Заказы

| Колонка           | Описание                                    | Тип         | Пример                                                  |
| ----------------- | ------------------------------------------- | ----------- | ------------------------------------------------------- |
| `id`              | Уникальный идентификатор заказа             | str         | `ORD123456`                                             |
| `dt`              | Дата заказа                                 | str / float | `2024-05-01`, `2024.5`, `1924-01-01`                    |
| `barista`         | Имя баристы                                 | str / None  | `Ирина`                                                 |
| `client`          | Идентификатор клиента                       | str / None  | `CL1234`                                                |
| `products`        | Список ID товаров в заказе                  | list\[str]  | `["P001", "P014"]`                                      |
| `dop`             | JSON с инф. по напиткам: молоко, лед, сироп | dict / None | `{"milk": "кокосовое", "ice": true, "syrup": "малина"}` |
| `amt`             | Итоговая стоимость заказа                   | float       | `380.0`                                                 |
| `discount`        | Применена ли скидка                         | bool / str  | `1`, `0`, `да`, `нет`                                   |
| `session`         | ID сессии (null — заказ в кофейне)          | str / None  | `SESS456`                                               |
| `order_placement` | Где сделан заказ: приложение или кофейня    | str         | `приложение` / `кофейня`                                |

---

### 2. Таблица: `sessions` — Сессии в приложении

| Колонка      | Описание                      | Тип        | Пример                |
| ------------ | ----------------------------- | ---------- | --------------------- |
| `id`         | Уникальный ID сессии          | str        | `SESS456`             |
| `start_dttm` | Время начала сессии           | datetime   | `2024-10-15 08:32:00` |
| `end_dttm`   | Время окончания сессии        | datetime   | `2024-10-15 08:37:45` |
| `client`     | ID клиента                    | str / None | `CL1234`              |
| `add`        | Добавлялся ли товар в корзину | bool / str | `1`, `0`, `да`, `нет` |
| `purchase`   | Совершена ли покупка          | bool / str | `1`, `0`, `да`, `нет` |

---

### 3. Таблица: `products` — Товары

| Колонка | Описание                               | Тип   | Пример                 |
| ------- | -------------------------------------- | ----- | ---------------------- |
| `id`    | Уникальный ID товара                   | str   | `P014`                 |
| `food`  | Флаг еды (True — еда, False — напиток) | bool  | `True`                 |
| `name`  | Название товара                        | str   | `Капучино`, `Круассан` |
| `cost`  | Стоимость товара                       | float | `180.0`                |

---

### 4. Таблица: `users` — Пользователи

| Колонка | Описание                                     | Тип | Пример           |
| ------- | -------------------------------------------- | --- | ---------------- |
| `id`    | Уникальный ID пользователя                   | str | `CL1234`         |
| `os`    | Операционная система устройства пользователя | str | `iOS`, `Android` |
