#### 2. Почему статистика важна для продуктовой аналитики? (3 мин)  

- *Data-driven подход: принятие решений на основе данных* <br>
   Статистика помогает заменить догадки и интуицию на решения, обоснованные цифрами. Например, вы можете определить, какая функция продукта чаще всего используется, а какая приносит наибольший доход продукту.  
- *Проверка гипотез и A/B тестирование*  <br>
   Статистика позволяет проверить, действительно ли изменения (например, новый дизайн) улучшают метрики продукта, или наблюдаемая разница случайна.  
- *Выявление зависимостей в данных* <br>
   Статистика позволяет находить взаимосвязи между метриками, например, как время, проведенное в приложении, связано с вероятностью покупки. Методы корреляции и регрессии помогают понять, что влияет на ключевые показатели   
- *Прогнозы и оптимизация*  <br>
   Статистика помогает строить прогнозы (например, роста аудитории) и приоритизировать задачи, чтобы сосредоточиться на самых важных улучшениях.  

**Итог**: Статистика — это ключевой инструмент для понимания данных, поиска взаимосвязей и принятия решений, которые делают продукт лучше.

# Основы теории вероятностей (45 мин)

---

#### 1. Пространство элементарных событий (2 мин)  
- ***Элементарное событие*** — один из возможных исходов эксперимента.  
  *Пример*: при подбрасывании шестигранного кубика элементарные события — значения, которые могут выпасть: 1, 2, 3, 4, 5, 6.

- ***Пространство элементарных событий*** $\Omega$ — множество всех возможных исходов эксперимента.  
  *Пример*: $\Omega = \{1, 2, 3, 4, 5, 6\}$.  

---

#### 2. События и их операции (3 мин)  
- ***Событие*** — это подмножество пространства элементарных событий: $A \subseteq \Omega.$  
  *Пример*: при броске кубика событие $A$ — "выпало четное число" ($A = \{2, 4, 6\}$).  
  
  
- **Операции с событиями**:  
  1. ***Объединение*** $A \cup B$ — событие, происходящее, если происходит $A$, $B$, или оба.  
     *Пример*: $A = \{1, 2, 3\}, B = \{3, 4, 5\}, A \cup B = \{1, 2, 3, 4, 5\}$.  
  2. ***Пересечение*** $A \cap B$ — событие, происходящее, если одновременно происходят $A$ и $B$.  
     *Пример*: $A \cap B = \{3\}$.  
  3. ***Дополнение*** $\overline{A}$ — событие, когда $A$ не происходит.  
     *Пример*: $\overline{A} = \{4, 5, 6\}$.
  4. ***Разность*** $A \setminus B$ — событие, когда $A$ происходит, а $B$ не происходит.  
     *Пример*: $A \setminus B = A \cap \overline{B} = \{1, 2\}$. 
  
  
- ***Пространство событий*** $\mathcal{F}$ — подмножество множества всех событий. Чаще всего будем рассматривать в качестве пространства событий множество всех событий.

#### 3. Что такое вероятность? (15 мин)  

- ***Вероятность*** — это мера уверенности в том, что случайное событие произойдет.  
  Пример: вероятность выпадения 1 на шестигранном кубике 1/6.  
  
  
- **Аксиомы вероятности**:
  1. Неотрицательность: $P(A) \geq 0.$
  2. Аддитивность: если $A_1$ и $A_2$ не пересекаются, то вероятность объединения равна сумме вероятностей: $A_1 \cap A_2 = \emptyset \ \rightarrow P(A_1 \cup A_2) = P(A_1) + P(A_2)$.  
     *Пример*: $P(\text{выпало 1 или 2}) = P(1) + P(2) = 1/3$.
  3. Вероятность всех событий равна 1: $P(\Omega) = 1$.  
     *Пример*: $P(\text{выпадет какое-то число от 1 до 6}) = P(1) + P(2) + P(3) + P(4) + P(5) + P(6) = 1$.
  
  
- **Свойства вероятности**:
  1. Вероятность пустого события 0: $P(\emptyset) = 0.$
  2. Вероятность большего события больше: если $A_1 \subseteq A_2$, то $P(A_1) \leq P(A_2)$.  
     *Пример*: $P(\text{выпало 1 или 2}) \leq P(\text{выпало 1 или 2 или 3})$.
  3. Вероятность дополнения — это 1 минус вероятность события: $P(\overline{A}) = 1 - P(A)$.  
     *Пример*: $P(\text{выпадет четное число}) = 1 - P(\text{выпадет нечетное число}) = 1/2$.
  4. Вероятность объединения через вероятности событий и вероятность пересечения: $P(A_1 \cup A_2) = P(A_1) + P(A_2) - P(A_1 \cap A_2).$  
     *Пример*: $P(\text{выпадет 1 или 2} \cup \text{выпадет 2 или 3}) = P(\text{выпадет 1 или 2}) + P(\text{выпадет 2 или 3}) - P(\text{выпадет 2})$.
     
- **Задачка на вероятности**:


**Задача: Домашние животные**

У класса из 30 учеников провели опрос: есть ли у них собака или кошка. Вот что известно:  
- 12 учеников сказали, что у них есть собака.  
- 18 учеников сказали, что у них есть кошка.  
- 5 учеников сказали, что у них есть и собака, и кошка.

*Вопросы:*
1. Какова вероятность, что случайно выбранный ученик имеет либо собаку, либо кошку?  
2. Какова вероятность, что случайно выбранный ученик не имеет ни собаки, ни кошки?


*Решение:*
1. Вероятность того, что у ученика есть либо собака, либо кошка, вычисляется с использованием формулы включений-исключений (свойство 4):

  $$P(\text{собака или кошка}) = P(\text{собака}) + P(\text{кошка}) - P(\text{собака и кошка}).$$

   Подставляем значения вероятностей (доли учеников):  
   $$P(\text{собака}) = \frac{12}{30}, \quad P(\text{кошка}) = \frac{18}{30}, \quad P(\text{собака и кошка}) = \frac{5}{30}.$$

   Тогда:  
   $$P(\text{собака или кошка}) = \frac{12}{30} + \frac{18}{30} - \frac{5}{30} = \frac{25}{30}.$$

   Ответ: вероятность, что у ученика есть либо собака, либо кошка, равна $\frac{25}{30} \approx 0.833$ (или 83.3%).

2. Вероятность отсутствия животного:  
   Если у ученика нет ни собаки, ни кошки, то это дополнительное событие к «собака или кошка». Поэтому:

   $$P(\text{нет животных}) = 1 - P(\text{собака или кошка}).$$

   Подставляем:  
   $$P(\text{нет животных}) = 1 - \frac{25}{30} = \frac{5}{30} \approx 0.167 \, (\text{или 16.7\%}).$$
   
*Ответ:* вероятность наличия хотя бы одного животного: $\frac{5}{6},$ вероятность отсутствия обоих животных: $\frac{1}{6}.$


*Дополнение:*

Эту задачу можно еще проиллюстрировать с помощью простого рисунка: круга для тех, у кого есть собаки, и круга для тех, у кого есть кошки, с пересечением, показывающим тех, у кого есть оба питомца. Это помогает визуализировать идею включений-исключений.

#### 4. Условная вероятность (5 мин)  
- ***Вероятность события $A$, если известно, что произошло событие $B$***:  
  $$P(A|B) = \frac{P(A \cap B)}{P(B)}, \, P(B) > 0.$$  
  *Пример*: вероятность, что выпадет 6, если известно, что выпало четное число.  
  
  
- **Независимость событий**: события $A$ и $B$ ***независимы***, если $P(A \cap B) = P(A) \cdot P(B)$.  
  *Пример*: вероятность, что выпадет 6, если известно, что в прошлый раз выпало 3.  
  
- **Задачка на условные вероятности**:  

**Задача: Игральные кости**  
Бросают две игральные кости. Известно, что сумма выпавших очков равна 8. 

*Вопрос:*
- Какова вероятность того, что на первой кости выпало число 5?  

*Решение:*  

1. Пространство элементарных событий:  
   Всего возможных пар исходов при броске двух костей — $6 \times 6 = 36$.  

2. Условие:  
   Рассматриваем только те исходы, где сумма очков равна 8:  
   $$(2, 6), (3, 5), (4, 4), (5, 3), (6, 2).$$  
   Таким образом, всего таких исходов 5.  

3. Интересующее событие:  
   На первой кости выпало 5. Подходящий исход — $(5, 3)$.  

4. Условная вероятность:  
   По формуле условной вероятности:  
   $$P(A|B) = \frac{P(A \cap B)}{P(B)}.$$  
   Вероятность события $B$ (сумма равна 8):  
   $$P(B) = \frac{\text{Число подходящих исходов}}{\text{Общее число исходов}} = \frac{5}{36}.$$  
   Вероятность события $A \cap B$ (первая кость — 5, сумма — 8):  
   $$P(A \cap B) = \frac{1}{36}.$$  
   Условная вероятность:  
   $$P(A|B) = \frac{P(A \cap B)}{P(B)} = \frac{\frac{1}{36}}{\frac{5}{36}} = \frac{1}{5}.$$  

---

*Ответ:*  
Вероятность того, что на первой кости выпало число 5, если известно, что сумма равна 8, равна $\frac{1}{5}$.

#### 5. Случайные величины и распределения (5 мин)  
- ***Случайная величина*** $X$ — числовая характеристика исхода эксперимента.  
  *Пример*: сумма чисел при 2 бросках кубика.  
  
  
- ***Распределение случайной величины:*** 
  - ***Дискретные случайные величины***:  
    Распределение задается вероятностями отдельных значений ($P(X = x)$).  
    *Пример*: бросок кубика: $P(X = i) = 1/6, \, i = 1, 2, \dots, 6$.  
  - ***Непрерывные случайные величины***:  
    Задаются функцией плотности вероятности $p(x)$, где вероятность лежать в интервале:  
    $$P(a \leq X \leq b) = \int\limits_a^b p(x) \, dx.$$
    *Пример*: генерация случайного числа на отрезке $[0, 1]$: $p(x) = 1, \, x \in [0, 1]$ или $p(x) = 0$ если $x \notin [0, 1]$.
    
- ***Функция распределения случайной величины:*** вероятность получить значение $x$ или меньше: $$F(x) = P(X \leq x).$$
  - ***Дискретные случайные величины***:  
    $$F(x) = \sum_{y \leq x} P(y).$$
    *Пример*: бросок кубика: $F(3) = P(\text{выпадет 3 или меньше}) = P(1) + P(2) + P(3) = 1/2$.  
  - ***Непрерывные случайные величины***:
    $$F(x) = \int\limits_{-\infty}^x p(y)\,dy.$$  
    *Пример*: $$F(0.5) = P(\text{сгенерированное число будет не больше 0.5}) = \int\limits_{0}^{0.5} 1\,dy = 1/2.$$

#### 6. Математическое ожидание, дисперсия и их свойства (15 мин)  
- ***Математическое ожидание*** $\mathsf{E}X$: среднее значение случайной величины.  
  $$\mathsf{E}X = \sum_x x \cdot P(X = x) \,\, \text{— для дискретной сл. величины} \quad \text{или} \\ \mathsf{E}X = \int\limits_{-\infty}^\infty x p(x) dx \,\, \text{— для непрерывной сл. величины}.$$  
  *Пример*: для подбрасывания кубика матожидание выпавшего числа: $\mathsf{E}X = 1/6 \cdot 1 + 1/6 \cdot 2 + 1/6 \cdot 3 + 1/6 \cdot 4 + 1/6 \cdot 5 + 1/6 \cdot 6 = 3.5$.  
  
  
- ***Дисперсия*** $\mathsf{D}X$: квадратичная мера разброса величины вокруг среднего.  
  $$\mathsf{D}X = \mathsf{E}[(X - \mathsf{E}X)^2].$$  

---
- **Свойства**:
  
    - **Математическое ожидание**:  
      1. ***Линейность***:  
         $$\mathsf{E}[aX + bY] = a\mathsf{E}X + b\mathsf{E}Y.$$  
         *Пример*: если $X$ — доход от продаж, а $Y$ — расходы, то матожидание прибыли $V$:  
         $$\mathsf{E}V = \mathsf{E}[X - Y] = \mathsf{E}X - \mathsf{E}Y.$$  
      2. Для константы $c$: $\mathsf{E}c = c$.  
      3. Для произведения **независимых** случайных величин:  
         $$\mathsf{E}[XY] = \mathsf{E}X\mathsf{E}Y.$$ 

    - **Дисперсия**:  
      1. Для суммы **независимых** случайных величин:  
         $$\mathsf{D}[X + Y] = \mathsf{D}X + \mathsf{D}Y.$$  
      2. Если $a$ — константа:  
         $$\mathsf{D}a = 0.$$
      3. Если $b$ — константа:  
         $$\mathsf{D}[bX] = b^2\mathsf{D}X.$$ 
      4. Дисперсия всегда неотрицательна: $$\mathsf{D}X \geq 0.$$  
      5. Связь с матожиданием квадрата: $$\mathsf{D}X = \mathsf{E}X^2 - (\mathsf{E}X)^2.$$

    - Связь с стандартным отклонением:  
      ***Стандартное отклонение*** — это квадратный корень из дисперсии:  
      $$\sigma(X) = \sqrt{\mathsf{D}X}.$$
      
      
**Задача: Игральные кости 2**  

Условие:  
Бросают игральную кость. Выпавшее число $X$ (очки на грани) случайным образом определяет приз в рублях, который получает игрок:  
- Если выпадает $1$, $2$ или $3$, то приз равен 10 рублей.  
- Если выпадает $4$, то приз равен 20 рублей. 
- Если выпадает $5$, то приз равен 30 рублей.
- Если выпадает $6$, то приз равен 50 рублей.  

Одна игра стоит 25 рублей.

*Вопрос:* 
1. Найти средний выигрыш в игре. Стоит ли играть в эту игру?  
2. Найти среднеквадратичное отклонение выигрыша.


*Решение:*  

1. Формируем случайную величину $X$, отвечающую за размер приза:  
   - При значениях $1$, $2$, $3$ (3 из 6 возможных): $P(X = 10) = \frac{3}{6}$.  
   - При значении $4$ (1 из 6 возможных): $P(X = 20) = \frac{1}{6}$.  
   - При значении $5$ (1 из 6 возможных): $P(X = 30) = \frac{1}{6}$.  
   - При значении $6$ (1 из 6 возможных): $P(X = 50) = \frac{1}{6}$.
   
2. Формализируем задачу: 
- средний выигрыш — это матожидание приза за вычетом стоимости игры
- если средний выигрыш будет больше нуля, то игра выгодная, а значит, в нее стоит сыграть, а если меньше нуля, то невыгодная, а значит, играть не стоит
- среднеквадратичное отклонение выигрыша — это стандартное отклонение. Так как игра имеет константную стоимость, то можно считать стандартное отклонение приза.

3. Средний выигрыш:  
   По формуле:  
   $$\mathsf{E}X = \sum_{x} x \cdot P(X = x).$$  
   Подставляем:  
   $$\mathsf{E}X = 10 \cdot \frac{3}{6} + 20 \cdot \frac{1}{6} + 30 \cdot \frac{1}{6} + 50 \cdot \frac{1}{6} = \frac{130}{6} \approx 21.67$$  
   
   Игра стоит 25 рублей, поэтому средний выигрыш будет $\mathsf{E}X - 25 \approx 21.67 - 25 = -3.33 < 0,$ а значит, в игру играть не стоит.
   

4. Среднеквадратичное отклонение приза:  
   По формуле:  
   $$\sigma(X) = \sqrt{\mathsf{E}X^2 - (\mathsf{E}X)^2}.$$  
   Сначала найдём $\mathsf{E}X^2$:  
   $$\mathsf{E}X^2 = \sum_{x}^n  x^2 \cdot P(X = x).$$  
   Подставляем:  
   $$\mathsf{E}X^2 = 10^2 \cdot \frac{3}{6} + 20^2 \frac{1}{6} + 30^2 \frac{1}{6} + 50^2 \frac{1}{6} = \frac{4100}{6}$$  

   Теперь cреднеквадратичное отклонение:  
   $$\sigma(X) = \sqrt{\frac{4100}{6} - \left(\frac{130}{6}\right)^2} = \sqrt{\frac{24600 - 16900}{36}} \approx 14.62.$$ 

*Ответ:*  
1. Cредний выигрыш будет примерно $-3.33 < 0,$ а значит, в игру играть не стоит.  
2. Среднеквадратичное отклонение выигрыша: $14.62$ рублей.

# Основные распределения (10 мин)

#### **1. Дискретные распределения**

1. **Бернуллиевское распределение**  
    - **Описание:** Дискретное распределение для события, которое имеет два исхода: "успех" (1) и "неудача" (0). 
    - **Пример 1:** Результат броска несимметричной монеты: орел (1) или решка (0). 
    - **Пример 2:** Пользователь завершил регистрацию (1) или нет (0).  
    - **Параметры:** Вероятность успеха $p$.  
    - **Плотность:** $P(X = 1) = p, \, P(X = 0) = 1 - p.$
---

2. **Биномиальное распределение**  
    - **Описание:** Распределение количества успехов в серии независимых экспериментов с двумя исходами.  
    - **Пример 1:** Сколько орлов выпадет при $n$ бросках несимметричной монеты?
    - **Пример 2:** Сколько пользователей завершат регистрацию, если начали регистрацию $n$?  
    - **Параметры:**  
      - $n$ — количество экспериментов.  
      - $p$ — вероятность успеха.  
    - **Плотность:** $P(X = k) = C_n^k p^k (1 - p)^{n - k}.$
  
---

3. **Геометрическое распределение**  
    - **Описание:** Описывает количество попыток до первого успеха в серии экспериментов с двумя исходами.  
    - **Пример 1:** Номер первого орла при бросках несимметричной монеты.
    - **Пример 2:** Сколько раз пользователь увидит рекламу до клика по ней?  
    - **Параметры:** Вероятность успеха $p$.
    - **Плотность:** $P(X = k) = p(1 - p)^k.$

---

4. **Пуассоновское распределение**  
    - **Описание:** Описывает число событий, происходящих за фиксированный интервал времени или пространства.  
    - **Пример:** Количество покупок, совершаемых пользователями в интернет-магазине в течение одного часа.
    - **Параметры:** Средняя частота событий $\lambda$.  
    - **Плотность:** $P(X = k) = \frac{\lambda^k}{k!}e^{-\lambda}$.

---

5. **Равномерное распределение**
    - **Описание:** Распределение для равновероятных событий. 
    - **Пример 1:** результат броска игральной кости.  
    - **Пример 2:** номер случайно выбранного пользователя из $n$ возможных.  
    - **Параметры:** $A$ — множество значений равномерного распределения.  
    - **Плотность:** $P(X = x) = 1/|A|$ для всех $x \in A$.

---

#### **2. Непрерывные распределения**

1. **Непрерывное равномерное распределение**  
    - **Описание:** Описывает случай, когда каждое значение внутри заданного интервала $[a, b]$ равновероятно. 
    - **Пример:** выбор случайной скидки для пользователя из интервала $[5\%, 20\%]$.  
    - **Параметры:** Интервал $[a, b]$. 
    - **Плотность** $p(x) = \frac{1}{b-a}$ если $x \in [a, b]$.  

---

2. **Нормальное распределение**  
    - **Описание:** Симметричное "колоколообразное" распределение, описывающее множество естественных процессов.  
    - **Пример:** рост/вес человека часто моделируется нормальным распределением.
    - **Параметры:**  
      - Математическое ожидание $a = \mathsf{E}X$ — среднее значение.  
      - Дисперсия $\sigma^2 = \mathsf{D}X$ — мера разброса. 
    - **Плотность:** $p(x) = \frac{1}{\sqrt{2\pi\sigma^2}}e^{-\frac{(x - a)^2}{2\sigma^2}}.$
  
--- 

3. **Экспоненциальное распределение**  
    - **Описание:** Описывает время между событиями в процессе, где они происходят с постоянной интенсивностью.  
    - **Пример:** Время до следующего визита пользователя на сайт.
    - **Параметры:** Параметр $\lambda$, обратный среднему времени между событиями.    
    - **Плотность:** $p(x) = \lambda e^{-\lambda x}$ для $x \geq 0$.

# Выборка и ген. совокупность. Точечные оценки (20 минут)

---

#### **1. Что такое выборка и зачем она нужна? (3 минуты)**

- ***Генеральная совокупность:*** все объекты или данные, которые мы хотим изучить.  
   *Пример:* все пользователи приложения.  
- ***Выборка:*** подмножество генеральной совокупности, которое используется для анализа.  
   *Пример*: из 1 млн пользователей изучаем поведение выбранных 10 000.  

- **Почему не взять всю генеральную совокупность?**  
  - Не можем собрать данные обо всех возможных наблюдениях и характеристиках.  
   *Пример*: не можем опросить всех пользователей приложения.
  - Достаточно затратно по времени и ресурсам: обработать данные по 10 000 пользователям быстрее, чем по 1 млн. 
  
**Вывод:** мы не всегда можем работать с каждым элементом, поэтому зачастую анализируем часть данных.

---

#### **2. Выборочное пространство и случайная выборка (3 минуты)**

- ***Случайная выборка:*** Каждый элемент генеральной совокупности имеет равные шансы быть выбранным. Это важно для объективности выводов.
   *Пример*: выбираем 10 000 пользователей приложения случайным образом.  
   **Замечание**: в дальнейшем, если не оговорено иное, **под выборкой мы будем понимать именно случайную выборку**, то есть, последовательность независимых и одинаково распределенных величин.
- ***Выборочное пространство:*** Это множество всех возможных выборок, которые можно получить из генеральной совокупности.  
   *Пример*: всевозможные комбинации из 10 000 пользователей приложения.
   
---

#### **3. Основные характеристики выборки (4 минуты)**

- **Параметры и точечные оценки:**  
  - ***Параметры:*** Характеристики всей генеральной совокупности, которые мы хотим узнать (например, средний возраст всех пользователей).  
  - ***Точечные оценки:*** Характеристики выборки, которые служат приближением параметров (например, средний возраст в выборке).  

- **Выборочные характеристики:**  
  1. ***Выборочное среднее:***  
     $$\overline{X} = \frac{1}{n} \sum_{i=1}^n X_i$$  
     Это оценка математического ожидания $\mathsf{E}X$.  

  2. ***Выборочная дисперсия:***  
     $$S^2 = \frac{1}{n-1} \sum_{i=1}^n (X_i - \overline{X})^2$$  
     Это оценка дисперсии $\mathsf{D}X$. 
     
  3. ***Выборочная квантиль:*** Значение, которое делит упорядоченные данные на части определённой пропорции.  
      - $\alpha$-квантиль ($\widehat{Q}_{\alpha}$): значение, ниже которого находится $\alpha \cdot 100\%$ наблюдений.  
      - Частный случай:  
        - Выборочная медиана $\widehat{Q}_{0.5}$ — выборочная 0.5-квантиль.  
      - Это оценка $\alpha$-квантиля распределения $Q_{\alpha}$: $$Q_{\alpha} = \min\{x | \, P(X \geq x) \geq \alpha\}.$$ 
      
### **Итог**  
Выборка — это инструмент для изучения генеральной совокупности, а статистические методы позволяют оценивать параметры, проверять гипотезы и строить прогнозы. Основные понятия, такие как ЦПТ, ЗБЧ и дельта-метод, обеспечивают надежность анализа данных, что делает их незаменимыми на практике.
